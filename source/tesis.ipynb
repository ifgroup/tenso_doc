```python
from math import ceil
import os
import json as json
import numpy as np
from tqdm import tqdm

from tenso.prototypes.heom import system_multibath
from tenso.prototypes.bath import gen_bcf
import matplotlib.pyplot as plt
```

```python
bath_vibr1 = gen_bcf(
        re_d=[3097.68], 
        width_d=[100.0],
        temperature=300,
        decomposition_method='Pade',
        n_ltc=1,
    )

bath_vibr2 = gen_bcf(
        re_d=[2425.8], 
        width_d=[100.5],
        temperature=300,
        decomposition_method='Pade',
        n_ltc=1,
    )

bath_solv = gen_bcf(
        re_d=[596.2],
        width_d=[50.0],
        temperature=300,
        decomposition_method='Pade',
        n_ltc=1,
)

N_sys = 3

sys_ops = []
end_time = 10000.0
dt = 1
for i in range(3):
    op_i = np.zeros((3, 3))
    op_i[i, i] = 1.0
    sys_ops.append(op_i) # The system operator of a site is a list of projective operators [|n><n|]
```

```python
#Define system operators

vibr_1 = np.array([
    [0.37, 0.34, 0], [0.34, 0.37, 0], [0, 0, 0]
], dtype=np.complex128)  # Vibrational mode 1

vibr_2 = np.array([
    [0, 0, 0], [0, 0.37, 0.34], [0, 0.34, 0.37]
], dtype=np.complex128)  # Vibrational mode 2

sys_ops.append(vibr_1)  # Add vibrational mode 1 operator
sys_ops.append(vibr_2)  # Add vibrational mode 2 operator


#Define system params

e1 = 15107.6
e2 = 23906.2

H = np.array([
    [0, 0, 0],
    [0, e1, 0],
    [0, 0, e2]
], dtype=np.complex128)


wfn = np.array([1.0, 0.0, 0.0], dtype=np.complex128)  # Initial state of the system
```

```python
out = "prueba_tesis"
end_time = 100.0
dt = 1
propagator = system_multibath(
    fname=out,
    init_rdo=np.outer(wfn, wfn.conj()),
    sys_ham=H,
    sys_ops=sys_ops,
    bath_correlations=[bath_solv, bath_solv, bath_solv, bath_vibr1, bath_vibr2],
    end_time=end_time,
    step_time=dt,
    dim=20
)

progress_bar = tqdm(propagator, total=ceil(end_time / dt)) #Show the progress bar
for _t in (progress_bar):
    progress_bar.set_description(f'@{_t:.2f} fs')
```
